```python
def get_word_vectors(tokenizer, embedding_dim=8):
    """
    Returns a dict mapping each word in tokenizer.word_index to a vector,
    using a (randomly-initialized, untrained) Keras Embedding layer.
    """
    import numpy as np
    from tensorflow.keras.layers import Embedding

    vocab_size = len(tokenizer.word_index) + 1  # +1 for padding index 0
    embedding_layer = Embedding(input_dim=vocab_size, output_dim=embedding_dim)
    embedding_layer.build((None,))  # initialize weights
    weights = embedding_layer.get_weights()[0]

    return {word: weights[idx] for word, idx in tokenizer.word_index.items()}
```

```python
word2idx = simple_tokenizer.word_index
vocab_size = len(word2idx)

def get_word_vectors(tokenizer):
    word2idx = tokenizer.word_index
    vocab_size = len(word2idx)
    
    # build identity matrix -> each row is a one-hot vector
    one_hot_matrix = np.eye(vocab_size, dtype=int)
    
    df = pd.DataFrame(one_hot_matrix, columns=range(1, vocab_size + 1))
    df.insert(0, 'word', word2idx.keys())
    df.index = range(1, vocab_size + 1)
    return df

get_word_vectors(simple_tokenizer)
```

```python
def get_sentence_vectors(sentences, tokenizer, mode="binary"):
    matrix = tokenizer.texts_to_matrix(sentences, mode=mode)
    df = pd.DataFrame(matrix)
    df.drop(columns=0, inplace=True)
    df.columns = tokenizer.word_index
    return df
```

```python
get_sentence_vectors(sentences, tokenizer, mode="tfidf")
```
