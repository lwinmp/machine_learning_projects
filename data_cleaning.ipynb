```python
#ITEM_BASED
data_ib = df_wide.copy()
```

```python
data_ib.head()
```

```python
data_ib = data_ib.reset_index()
```

```python
data_ib.head()
```

```python
data_ib = data_ib.drop('Person', axis=1)
```

```python
data_ib.head()
```

```python
data_ibs = pd.DataFrame(index=data_ib.columns, columns=data_ib.columns)
```

```python
data_ibs
```

```python
for i in range(0, len(data_ibs.columns)):
    for j in range(0, len(data_ibs.columns)):
        data_ibs.iloc[i,j] = 1 - cosine(data_ib.iloc[:,i], data_ib.iloc[:,j])
```

```python
data_ibs.head()
```

```python
data_neighbours = pd.DataFrame(index=data_ibs.columns, columns=range(1,11))
```

```python
for i in range(0, len(data_ibs.columns)):
    data_neighbours.iloc[i,:10] = data_ibs.iloc[0:,i].sort_values(ascending=False)[:10].index
```

```python
data_neighbours
```

```python
#End of Item-based collaborative filtering
```

```python
#Start of User-based collaborative filtering
```

```python
def getScore(history, similarities):
    return sum(history*similarities)/ sum(similarities)
```

```python
data_sim1 = df_wide.reset_index()
```

```python
data_sim1.head()
```

```python
data_sim = pd.DataFrame(index=data_sim1.index, columns=data_sim1.columns)
data_sim.iloc[:,:1] = data_sim1.iloc[:,:1]
```

```python
data_sim.head()
```

```python
data_sim12 = data_sim1.iloc[:500,:]
```

```python
data_sims11 = data_sim.iloc[:500,:]
```

```python
for i in range(len(data_sims11.index)):
    for j in range(1, len(data_sims11.columns)):
        user = data_sims11.index[i]
        product = data_sims11.columns[j]

        if data_sim12.iloc[i, j] == 1:
            data_sims11.iloc[i, j] = 0
        else:
            product_top_names = data_neighbours.loc[product].iloc[1:10]
            product_top_sims = data_ibs.loc[product].sort_values(ascending=False).iloc[1:10]
            user_purchases = data_ib.loc[user, product_top_names]

            data_sims11.iloc[i, j] = getScore(user_purchases, product_top_sims)
```
