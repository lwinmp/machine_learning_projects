```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

```python
url = "https://www.tunyat.com/"
response = requests.get(url)
print(response.status_code)
print(response.url)
print(response.headers)
```

```python
print(response.text[:2000])
```

```python
soup = BeautifulSoup(response.text, "html.parser")
print(soup.title)
```

```python
print(soup.title.get_text(strip=True))
```

```python
for child in soup.body.children:
    if child.name:
        print(child.name)
```

```python
headings = soup.find_all(["h1", "h2", "h3", "h4"])
print("Number of headings:", len(headings))

for heading in headings:
    print(heading.name, heading.get_text(" ", strip=True))
```

```python
heading_data = []
for heading in headings:
    heading_data.append(
        {
            "tag": heading.name,
            "text": heading.get_text("", strip=True)
        }
    )
heading_data
```

```python
df = pd.DataFrame(heading_data)
df
```

```python
links = soup.find_all("a")
link_data = []

for link in links:
    text = link.get_text("", strip=True)
    href = link.get("href")

    if text or href:
        link_data.append({
            "text": text,
            "url": href
        })
links_df = pd.DataFrame(link_data)
links_df
```

```python
links_clean = links_df.drop_duplicates()

links_clean
```

```python
links_unique = (
    links_df[
        links_df["url"].str.startswith(("http://", "https://"), na=False)
    ]
    .drop_duplicates(subset="url")
    .reset_index(drop=True)
)

links_unique
```

```python
internal_links = links_unique[
    links_unique["url"].str.startswith("https://www.tunyat.com")
]
```

```python
internal_links = internal_links[["text", "url"]].reset_index(drop=True)

internal_links
```

```python
internal_links.to_csv(
    "tunyat.csv",
    index=False
)
```
