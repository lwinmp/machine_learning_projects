```python
df = df.dropna(subset=['rating_count'])
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df['rating'], bins=10, kde=True)
plt.title("Rating Distribution")
plt.xlabel("Rating")
plt.ylabel("Frequence")
plt.show()
```

```python
plt.figure(figsize=(45,20))
sns.histplot(df['discount_percentage'], bins=20, kde=True)
plt.title("Discount Distribution")
plt.show()
```

```python
plt.figure(figsize=(12,6))
sns.histplot(df['actual_price'], bins=30, kde=True)
plt.title("Price Distribution")
plt.show()
```

```python
df.sort_values(by='actual_price').tail()
```

```python
df[['actual_price', 'discounted_price']].info()
```

```python
df = df[df["rating"] != "|"]
```

```python
df['rating'] = df['rating'].astype(float)
```

```python
df.columns
```

```python
df['rating_count'] = df['rating_count'].str.replace(',', '').astype(float)
```

```python
df.head(5)
```

```python
median_count = df['rating_count'].median()
df['rating_count'] = df['rating_count'].fillna(median_count)
```

```python
df.head(5)
```

```python
df['discounted_price'] = df['discounted_price'].str.replace('₹', '').str.replace(',', '').astype(float)
df['actual_price'] = df['actual_price'].str.replace('₹', '').str.replace(',', '').astype(float)
```

```python
df['discount_percentage'] = df['discount_percentage'].str.replace('%', '').astype(float)
```

```python
df.head(5)
```

```python
df.isnull().sum()
```

```python
df.info()
```

```python
df['rating'].unique()
```

```python
df['rating_count'].unique()
```

```python
df['rating_count'] = df['rating_count'].str.replace(',', '').astype(float)
```

```python
df['rating_count'].unique()
```

```python
df['rating'] = pd.to_numeric(df['rating'], errors='coerce')
```

```python
df['rating'].unique()
```

```python
df['rating_count'] = pd.to_numeric(df['rating_count'], errors='coerce')
```

```python
correlation = df['rating'].corr(df['rating_count'])
print(f"Mathematical Correlation between Rating and Sales Proxy: {correlation:.4f}")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid')
fig, axes = plt.subplots(1, 2, figsize=(10,6))
sns.histplot(df['rating'], bins=30, kde=True, color='blue', ax=axes[0])
axes[0].set_title('Distribution of Product Rating', fontsize=12)
axes[0].set_xlabel('Rating (1-5 Stars)')
axes[0].set_ylabel('No: of Products')

sns.scatterplot(x='rating', y='rating_count', data=df, alpha=0.5, color='orange', ax=axes[1])
axes[1].set_title('Do higher ratings mean more sales?', fontsize=12)
axes[1].set_xlabel('Product Rating')
axes[1].set_ylabel('Rating Count')

plt.tight_layout()
plt.show()

correlation = df['rating'].corr(df['rating_count'])
print(f"Mathematical Correlation between Rating and Sales Proxy: {correlation:.4f}")
```
