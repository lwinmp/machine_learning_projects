```python
import numpy as np
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer, TfidfTransformer
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv("spamhamdata.csv", encoding="ISO-8859-1", sep="\t", header=None, names=["type", "text"])
df.head()
```

```python
df.info()
```

```python
df.columns
```

```python
import re
stop_words = []
with open("stop.txt") as f:
    stop_words = f.read()

stop_words = stop_words.split("\n")
```

```python
def clean_text(i):
    i = re.sub("[^A-Za-z" "]+", " ", i).lower()
    i =  re.sub("[0-9" "]+", " ",i)
    w = []
    for word in i.split(" "):
        if len(word)>3:
            w.append(word)
        return (" ".join(w))
```

```python
df.text = df.text.apply(clean_text)
df
```

```python
df = df.loc[df.text != " ",:]
```

```python
def split_words(i):
    return [word for word in i.split(" ")]
```

```python
X_train, X_test = train_test_split(df, test_size=0.2)
```

```python
mail_count = CountVectorizer(analyzer=split_words).fit(df.text)
```

```python
mail_matrix = mail_count.transform(df.text)
```

```python
mail_matrix.shape
```

```python
train_mail_matrix = mail_count.transform(X_train.text)
```

```python
train_mail_matrix.shape
```

```python
test_mail_matrix = mail_count.transform(X_test.text)
```

```python
test_mail_matrix.shape
```

```python
from sklearn.naive_bayes import MultinomialNB as MB

model_1 = MB()
model_1.fit(train_mail_matrix, X_train.type)

train_pred = model_1.predict(train_mail_matrix)
train_acc = np.mean(train_pred==X_train.type)
print("Train Accuracy:", train_acc)

test_pred = model_1.predict(test_mail_matrix)
test_acc = np.mean(test_pred==X_test.type)
print("Test Accuracy:", test_acc)
```

```python
from sklearn.naive_bayes import GaussianNB as GB

model_2 = GB()

#convert matrix to dense
train_dense = train_mail_matrix.toarray()
test_dense = test_mail_matrix.toarray()

model_2.fit(train_dense, X_train.type)

train_pred = model_2.predict(train_dense)
train_acc = np.mean(train_pred==X_train.type)
print("Train Accuracy:", train_acc)

test_pred = model_2.predict(test_dense)
test_acc = np.mean(test_pred==X_test.type)
print("Test Accuracy:", test_acc)
```

```python
from sklearn.linear_model import LogisticRegression as LR
import numpy as np

model_3 = LR()
model_3.fit(train_mail_matrix, X_train.type)

train_pred = model_3.predict(train_mail_matrix)
train_acc = np.mean(train_pred == X_train.type)
print("Train Accuracy:", train_acc)

test_pred = model_3.predict(test_mail_matrix)
test_acc = np.mean(test_pred == X_test.type)
print("Test Accuracy:", test_acc)
```
