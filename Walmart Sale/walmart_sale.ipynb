```python
df['Date'] = pd.to_datetime(df['Date'], format='%d-%m-%Y')
```

```python
num_cols =  df.select_dtypes(include=['number']).columns.drop('Holiday_Flag')
extra_num_col = ['Holiday_Flag']
```

```python
num_cols
```

```python
print(f"Total Numerical Columns: {len(num_cols)}")
```

```python
for col in num_cols:
    fig, axes = plt.subplots(1, 2, figsize=(10,5))
    sns.histplot(data=df, x=col, kde=True, ax=axes[0], color='skyblue')
    axes[0].set_title(f'{col} Distribution')

    sns.histplot(data=df, x=col, ax=axes[1], color='lightcoral')
    axes[1].set_title(f'{col} Outliners')

    plt.tight_layout()
    plt.show()
```
