```python
import pandas as pd
from scipy.spatial.distance import cosine
```

```python
df = pd.read_csv('groceries.csv')
df.head()
```

```python
df["Quantity"] = 1
```

```python
df.head()
```

```python
len(pd.unique(df.item))
```

```python
dataWide = df.pivot(
    index="Person", 
    columns="item", 
    values="Quantity")
dataWide.head()
```

```python
dataWide.fillna(0, inplace=True)
```

```python
dataWide.head()
```

```python
#Item-based Collaborative Filtering
#In item based collaborative filtering we do not care about the user column
```

```python
data_id = dataWide.copy()
data_id.head()
```

```python
data_id = data_id.reset_index()
data_id.head()
```

```python
data_id = data_id.drop("Person", axis=1)
```

```python
data_id.head()
```

```python
# Create a placeholder dataframe listing item vs. item
data_ids = pd.DataFrame(index=data_id.columns, columns=data_id.columns)
```

```python
data_ids.head()
```

```python
data_ids = pd.DataFrame(index=data_id.columns, columns=data_id.columns)
```

```python
#consine similarity
```

```python
for i in range(0, len(data_ids.columns)):
    for j in range(0, len(data_ids.columns)):
        data_ids.iloc[i,j] = 1 - cosine(data_id.iloc[:,i], data_id.iloc[:,j])
```
